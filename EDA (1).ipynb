## What are the Top 5 most installed Apps in Each popular Categories ??

```python
df = pd.read_csv('/content/Top 1000 technology companies.csv')
df_copy = df.copy()
def convert_to_billions(value):
    value = str(value).replace('$', '').replace(',', '').strip()
    multipliers = {'T': 1000, 'B': 1, 'M': 0.001, 'K': 0.000001}

    if value[-1:] in multipliers:
        return float(value[:-1].strip()) * multipliers[value[-1]]

    return float(value) / 1_000_000_000

df_copy['Market Cap'] = df_copy['Market Cap'].apply(convert_to_billions)

dfa = df_copy[['Industry', 'Company', 'Market Cap']].copy()
dfa = dfa.sort_values('Market Cap', ascending=False)

industries = [
    'Consumer Electronics',
    'Software—Infrastructure',
    'Semiconductors',
    'Software—Application'
]

sns.set_context("talk")
sns.set_style("darkgrid")

plt.figure(figsize=(22, 16))

for i, industry in enumerate(industries):
    df2 = dfa[dfa['Industry'] == industry]
    df3 = df2.head(5)

    plt.subplot(2, 2, i + 1)
    sns.barplot(data=df3, x='Market Cap', y='Company')
    plt.xlabel('Market Capitalization (USD Billions)')
    plt.ylabel('')
    plt.title(industry, size=20)

plt.tight_layout()
plt.show()
```

## Insights
- Most popular game is Subway Surfers.
- Most popular communication app is Hangouts.
- Most popular productivity app is Google Drive.
- Most popular social app is Instagram.
